## 5. Export & modelling
Median / mode imputation lives inside the pipeline, so it is fit on training data only (no leakage).
Target is `log1p(SalePrice)`; metrics are reported on **both** the log scale and in dollars.

```python
df.to_csv("clean_train.csv", index=False)
```

```python
from sklearn.model_selection import train_test_split, KFold, cross_validate
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.dummy import DummyRegressor
from sklearn.linear_model import RidgeCV
from sklearn.ensemble import RandomForestRegressor, HistGradientBoostingRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.inspection import permutation_importance

SEED = 42
drop_cols = ["SalePrice", "Id", "SaleDate", "ListingAgent"]   # target, id, raw date, high-cardinality id-like
X = df.drop(columns=drop_cols, errors="ignore").copy()
y = np.log1p(df["SalePrice"])

num = X.select_dtypes(include="number").columns.tolist()
cat = [c for c in X.columns if c not in num]
X[cat] = X[cat].astype(object)          # KitchenQual is 'string' dtype; keep all categoricals uniform

# split FIRST; holdout is touched only once, at the end
Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=0.2, random_state=SEED)

def make_pre(scale=False):
    num_steps = [("imp", SimpleImputer(strategy="median"))]
    if scale:
        num_steps.append(("sc", StandardScaler()))
    return ColumnTransformer([
        ("num", Pipeline(num_steps), num),
        ("cat", Pipeline([
            ("imp", SimpleImputer(strategy="most_frequent")),
            ("oh", OneHotEncoder(handle_unknown="ignore", sparse_output=False)),
        ]), cat),
    ])

models = {
    "Dummy (mean)":     Pipeline([("pre", make_pre()), ("m", DummyRegressor(strategy="mean"))]),
    "Ridge":            Pipeline([("pre", make_pre(scale=True)), ("m", RidgeCV(alphas=np.logspace(-2, 3, 20)))]),
    "RandomForest":     Pipeline([("pre", make_pre()), ("m", RandomForestRegressor(300, min_samples_leaf=5, n_jobs=-1, random_state=SEED))]),
    "HistGradBoosting": Pipeline([("pre", make_pre()), ("m", HistGradientBoostingRegressor(learning_rate=0.05, max_iter=300, max_depth=4, l2_regularization=1.0, early_stopping=True, random_state=SEED))]),
}
```

### 5.1 Baseline + 5-fold cross-validation
The dummy model sets the floor: any real model must beat it. CV is run on the training split only.

```python
def mae_usd(est, X_, y_):
    return mean_absolute_error(np.expm1(y_), np.expm1(est.predict(X_)))

cv = KFold(n_splits=5, shuffle=True, random_state=SEED)
scoring = {"R2_log": "r2", "RMSE_log": "neg_root_mean_squared_error", "MAE_usd": mae_usd}

rows = {}
for name, pipe in models.items():
    r = cross_validate(pipe, Xtr, ytr, cv=cv, scoring=scoring, n_jobs=1)
    rows[name] = {
        "R2_log":   r["test_R2_log"].mean(),
        "R2_std":   r["test_R2_log"].std(),
        "RMSE_log": -r["test_RMSE_log"].mean(),
        "MAE_usd":  r["test_MAE_usd"].mean(),
    }
cv_results = pd.DataFrame(rows).T.sort_values("RMSE_log")
cv_results.round(4)
```

### 5.2 Holdout evaluation (best CV model)

```python
best_name = cv_results.drop(index="Dummy (mean)").index[0]
best = models[best_name].fit(Xtr, ytr)
dummy = models["Dummy (mean)"].fit(Xtr, ytr)

def report(est, label):
    p_log = est.predict(Xte)
    p_usd, y_usd = np.expm1(p_log), np.expm1(yte)
    return {
        "model": label,
        "RMSE_log": np.sqrt(mean_squared_error(yte, p_log)),
        "R2_log":   r2_score(yte, p_log),
        "MAE_usd":  mean_absolute_error(y_usd, p_usd),
        "R2_usd":   r2_score(y_usd, p_usd),
    }

holdout = pd.DataFrame([report(dummy, "Dummy (mean)"), report(best, best_name)]).set_index("model")
print("Best by CV:", best_name)
holdout.round(4)
```

### 5.3 Permutation importance (holdout, original features)

```python
pi = permutation_importance(best, Xte, yte, n_repeats=10, random_state=SEED, n_jobs=1)
imp = (pd.Series(pi.importances_mean, index=Xte.columns)
         .sort_values(ascending=False).head(15)[::-1])
err = pd.Series(pi.importances_std, index=Xte.columns)[imp.index]

plt.figure(figsize=(8, 6))
plt.barh(imp.index, imp.values, xerr=err.values)
plt.xlabel("Drop in R² (log scale) when feature is shuffled")
plt.title(f"Permutation importance - {best_name}")
plt.tight_layout()
plt.show()
```

### 5.4 Residual analysis

```python
pred_log = best.predict(Xte)
resid = yte - pred_log

fig, ax = plt.subplots(1, 3, figsize=(16, 4.5))
ax[0].scatter(np.expm1(yte), np.expm1(pred_log), alpha=0.5, s=15)
lims = [min(np.expm1(yte).min(), np.expm1(pred_log).min()), max(np.expm1(yte).max(), np.expm1(pred_log).max())]
ax[0].plot(lims, lims, "r--")
ax[0].set(xlabel="Actual price", ylabel="Predicted price", title="Predicted vs actual")

ax[1].scatter(pred_log, resid, alpha=0.5, s=15)
ax[1].axhline(0, color="r", ls="--")
ax[1].set(xlabel="Predicted log price", ylabel="Residual (log)", title="Residuals vs predicted")

sns.histplot(resid, kde=True, ax=ax[2])
ax[2].set(xlabel="Residual (log)", title="Residual distribution")
plt.tight_layout()
plt.show()
```

### 5.5 Why the score is low
The features carry little signal for `SalePrice`. In the EDA, `GrLivArea` correlates only ~0.25 with price, `LotArea` ≈ 0 and `GarageArea` ≈ 0.06; only `OverallQual` shows a clear upward trend, and neighbourhood price levels do not follow a sensible ordering. This looks like noisy (possibly synthetic) data, so the ceiling is low whatever the algorithm: the honest comparison is against the dummy baseline above, not against an R² of 1.
Model choice matters little here; more reliable gains would come from better features (ordinal-encoded quality columns, flags for pool/garage/fireplace), not from tuning.
